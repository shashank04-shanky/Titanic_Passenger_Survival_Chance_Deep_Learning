```python
import pandas as pd
```

```python
df=pd.read_csv('huge_1M_titanic.csv')
```

```python
df=df.sample(10000)
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df=df.drop(columns=['Cabin','PassengerId','Name','Ticket','Age'])
```

```python
df.head()
```

```python
df['Embarked'].value_counts()
```

### DATA CLEANING

```python
df['Embarked']=df['Embarked'].replace({'S':'Southampton','C':'Cherbourg','Q':'Queenstown'})
```

```python
df.head()
```

```python
df.dropna(subset=['Embarked'],inplace=True)
```

```python
df.head()
```

```python
df['Embarked'].value_counts()
```

```python
df['Fare']=df['Fare'].astype('int')
```

```python
df.head()
```

## Feature Enconding

#### Encode - Gender -> Label encoder
#### Embarked -> One hote encoding

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
df['Sex']=le.fit_transform(df['Sex'])
```

```python
df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder(sparse_output=False)
Embarked=ohe.fit_transform(df[['Embarked']])
Embarked=pd.DataFrame(Embarked,columns=ohe.get_feature_names_out())
```

```python
df.head()
```

```python
Embarked
```

```python
df=pd.concat([df.drop(columns=['Embarked']),Embarked],axis=1)
```

```python
df.head()
```

### feature scaling

```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
```

```python
df.columns
```

```python
num_columns=['Pclass', 'SibSp', 'Parch', 'Fare']
```

```python
df[num_columns]=scale.fit_transform(df[num_columns])
```

```python
df.head()
```

```python
import pickle
```

```python
with open('label_encoder.pkl','wb') as f:
    pickle.dump(le,f)
```

```python
with open('onehot_encoder.pkl','wb') as f:
    pickle.dump(ohe,f)
```

```python
with open('scalar.pkl','wb') as f:
    pickle.dump(scale,f)
```

```python
df=df.dropna()
```

### train_test_split

```python
X=df.drop(columns='Survived')
```

```python
Y=df['Survived']
```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
x_train,x_valid,y_train,y_valid=train_test_split(x_train,y_train,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

### Model building

```python
%pip install tensorflow
```

```python
import tensorflow
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential([
    Dense(128, activation='relu', input_shape=(x_train.shape[1],)),
    Dense(64,activation='relu'),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.summary()
```

### Model compilation

```python
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.01)
```

```python
model.compile(optimizer=opt, loss='binary_crossentropy', metrics=['accuracy'])
```

```python
stopping_callback = EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
```

```python
model.fit(x_train, y_train, validation_data=(x_valid, y_valid), epochs=10, batch_size=32, callbacks=[stopping_callback])
```

```python
model.save('titanic_model.h5')
```
