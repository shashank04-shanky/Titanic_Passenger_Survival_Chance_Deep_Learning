```python
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import tensorflow as tf
```

```python
with open('label_encoder.pkl','rb') as file : 
    label_encoder=pickle.load(file)
```

```python
with open('onehot_encoder.pkl','rb') as file : 
    onehot=pickle.load(file)
```

```python
with open('scalar.pkl','rb') as file : 
    scaler=pickle.load(file)
```

```python
model=load_model('titanic_model.h5')
```

```python
user={'Pclass':3,'Sex':'male','SibSp':1,'Parch':0,'Fare':11,'Embarked':'Queenstown'}
```

```python
user=pd.DataFrame([user])
```

```python
user.head()
```

```python
user['Sex']=label_encoder.transform(user['Sex'])
```

```python
user
```

```python
embarked=onehot.transform(user[['Embarked']])
```

```python
embarked=pd.DataFrame(embarked,columns=onehot.get_feature_names_out())
```

```python
embarked
```

```python
user=pd.concat([user,embarked],axis=1).drop('Embarked',axis=1)
```

```python
user[['Pclass','SibSp','Parch','Fare']]=scaler.transform(user[['Pclass','SibSp','Parch','Fare']])
```

```python
user
```

```python
y=model.predict(user)
```

```python
y=y[0][0]
```

```python
y
```

```python
if y>0.5:
    print("The passenger is likely to survive.")
else:
    print("The passenger is likely to not survive.")
```
